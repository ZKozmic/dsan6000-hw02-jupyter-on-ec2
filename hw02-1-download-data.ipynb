# Download Hourly Wikipedia Event Data

```python
# Imports
from pathlib import Path
import boto3
```

```python
source_bucket = "dsan6000-wikipedia"
prefix = "hourly_parquet/"
data_dir = Path("data")
data_dir.mkdir(exist_ok=True)

s3 = boto3.client("s3")

response = s3.list_objects_v2(
    Bucket=source_bucket,
    Prefix=prefix
)

parquet_keys = [
    item["Key"]
    for item in response.get("Contents", [])
    if item["Key"].endswith(".parquet")
]

for key in parquet_keys:
    local_file = data_dir / Path(key).name
    s3.download_file(source_bucket, key, str(local_file))

print(f"Downloaded {len(parquet_keys)} parquet files")
```
