# Store Wikipedia Events in an S3 Data Lake

```python
# Imports
from pathlib import Path
import boto3
from botocore.exceptions import ClientError
```

```python
bucket_name = "dsan6000-tz297"
s3 = boto3.client("s3")

try:
    s3.create_bucket(Bucket=bucket_name)
    print(f"Created bucket: {bucket_name}")
except ClientError as error:
    if error.response["Error"]["Code"] == "BucketAlreadyOwnedByYou":
        print(f"Bucket already exists: {bucket_name}")
    else:
        raise

parquet_files = sorted(Path("data").glob("*.parquet"))

for local_file in parquet_files:
    object_key = f"wikipedia-hourly/{local_file.name}"
    s3.upload_file(str(local_file), bucket_name, object_key)

print(f"Uploaded {len(parquet_files)} parquet files")
```
